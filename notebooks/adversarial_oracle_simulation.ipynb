```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def simuliere_orakel_schutz(korrupte_knoten_prozent, rauschanteil, gamma):
    """
    Simuliert die stochastische Orakel-Integrität und das Gewichts-Decay
    gemäß der mathematischen Formel aus Artikel II.
    """
    zeit = np.linspace(0, 100, 200)

    # Basis-Integrität der ehrlichen Knoten (nahe 100%)
    ehrliche_gewichtung = 100.0 - (korrupte_knoten_prozent * 0.1)

    # Berechnung des Gewichts-Decays basierend auf Rauschen und Gamma
    # w_i(t+1) = w_i(t) * exp(-gamma * D_KL)
    rausch_faktor = (rauschanteil / 100.0) * (korrupte_knoten_prozent / 100.0)
    degradations_kurve = ehrliche_gewichtung * np.exp(-gamma * rausch_faktor * zeit)

    plt.figure(figsize=(10, 5))
    plt.plot(zeit, degradations_kurve, label="Integrität des Orakel-Netzwerks ($w_i$)", color="#00ffcc", linewidth=2.5)
    plt.axhline(y=50, color="red", linestyle=":", label="Kritische Stabilitäts-Schwelle (50%)")

    plt.title("Artikel II: Byzantinische Orakel-Integrität & Gewichts-Decay", fontsize=14, color="white", pad=15)
    plt.xlabel("Zeitverlauf t (System-Ticks)", fontsize=12, color="white")
    plt.ylabel("Netzwerk-Gewichtung / Daten-Validität", fontsize=12, color="white")
    plt.grid(True, linestyle="--", alpha=0.3)
    plt.ylim(-5, 110)
    plt.legend(loc="lower left")

    # Dunkles HP-v10 Theme
    fig = plt.gcf()
    fig.patch.set_facecolor('#1e1e1e')
    ax = plt.gca()
    ax.set_facecolor('#121212')
    ax.spines['bottom'].set_color('white')
    ax.spines['top'].set_color('white')
    ax.spines['right'].set_color('white')
    ax.spines['left'].set_color('white')
    ax.tick_params(colors='white')
    plt.show()

print("=== HP-v10: INTERAKTIVES DASHBOARD (ARTIKEL I)")
print("-" * 55)

# Drei Schieberegler zur Manipulation des Angriffs-Szenarios
interact(
    simuliere_orakel_schutz,
    korrupte_knoten_prozent=widgets.IntSlider(min=0, max=100, step=5, value=30, description="Korrupte Orakel %:"),
    rauschanteil=widgets.FloatSlider(min=0.0, max=100.0, step=5.0, value=50.0, description="Rausch-Stärke:"),
    gamma=widgets.FloatSlider(min=0.001, max=0.05, step=0.002, value=0.01, description="Gamma (Dämpfung):")
)
```
