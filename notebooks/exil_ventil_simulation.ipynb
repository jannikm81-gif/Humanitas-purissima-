```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

def simuliere_exil_ventil(lokale_bedrohung, signatur_gueltig):
    """
    Simuliert das Exil- & Dissidenten-Ventil aus Artikel V.
    Visualisiert den asymmetrischen Zustandstransfer basierend auf Ed25519-Validierung.
    """
    zeit = np.linspace(0, 10, 100)

    # Lokaler Zustandsraum bricht bei hoher Bedrohung ein
    lokaler_status = max(0, 100.0 - lokale_bedrohung * 18)

    # Exil-Transfer-Status berechnen
    if signatur_gueltig and lokale_bedrohung > 2.0:
        # Transfer gelingt: Payload wandert asymmetrisch in den sicheren Raum
        exil_raum_status = lokale_bedrohung * 18
        transfer_erfolgreich = True
    else:
        exil_raum_status = 0.0
        transfer_erfolgreich = False

    plt.figure(figsize=(10, 5))
    plt.bar(["Lokaler Knotenraum", "Exil-Zustandsraum (Safe Haven)"], [lokaler_status, exil_raum_status],
            color=["#ff3366", "#00ffcc"], alpha=0.8, width=0.5)

    # Text-Indikator für das Krypto-Ventil
    if transfer_erfolgreich:
        plt.text(0.5, 50, "🔓 Ed25519 VERIFIED:\nTRANSFER ERFOLGREICH",
                 color="#00ffcc", fontsize=12, fontweight="bold", ha="center",
                 bbox=dict(facecolor='#121212', edgecolor='#00ffcc', boxstyle='round,pad=1'))
    else:
        plt.text(0.5, 50, "🔒 WAITING FOR SIGNATURE\nODER LOKAL STABIL",
                 color="#ff3366", fontsize=12, fontweight="bold", ha="center",
                 bbox=dict(facecolor='#121212', edgecolor='#ff3366', boxstyle='round,pad=1'))

    plt.title("Artikel V: Exil- & Dissidenten-Ventil via Ed25519-Signaturen", fontsize=14, color="white", pad=15)
    plt.ylabel("Zustands-Integrität / Payload-Volumen", fontsize=12, color="white")
    plt.ylim(-5, 110)

    # Dunkles HP-v10 Theme
    fig = plt.gcf()
    fig.patch.set_facecolor('#1e1e1e')
    ax = plt.gca()
    ax.set_facecolor('#121212')
    ax.spines['bottom'].set_color('white')
    ax.spines['top'].set_color('white')
    ax.spines['right'].set_color('white')
    ax.spines['left'].set_color('white')
    ax.tick_params(colors='white')
    plt.show()

print("=== HP-v10: KRYPTOGRAFISCHE TRANSFER-STEUERUNG ===")
print("-" * 55)

# Interaktive Regler für Bedrohungsstufe und Krypto-Signatur
interact(
    simuliere_exil_ventil,
    lokale_bedrohung=widgets.FloatSlider(min=0.0, max=5.0, step=0.5, value=0.0, description="Bedrohung:"),
    signatur_gueltig=widgets.Checkbox(value=True, description="Ed25519 Signatur gültig")
)
```
